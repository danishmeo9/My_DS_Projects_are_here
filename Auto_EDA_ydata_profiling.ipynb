```python
# Import necessary libraries
import pandas as pd
from pandas_profiling import ProfileReport

# Load your dataset
data = pd.read_excel('./data/Usage of Technology in ASRM Program (Responses).xlsx') # Replace 'your_dataset.csv' with your dataset file path

# Perform data profiling
profile = ProfileReport(data, title='Pandas Profiling Report', explorative=True)

# Generate the report
profile.to_file('data_profile_report.html')  # You can change the file name as needed

# Display the report in a Jupyter Notebook (optional)
# profile.to_notebook_iframe()

# Alternatively, you can display the report in the Jupyter Notebook if you're working in a Jupyter environment. Uncomment the 'to_notebook_iframe()' line if you want to use this option.

# To view the generated report, open the HTML file in a web browser or within your Jupyter Notebook (if you used the 'to_notebook_iframe()' option).

# Note: Make sure to install the required libraries using 'pip install pandas pandas-profiling' before running this code.

```
